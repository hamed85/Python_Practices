# **1) What is Apache Spark?**

Apache Spark is a powerful open-source distributed computing framework designed for processing and analyzing large-scale datasets efficiently. It enables users to distribute both data and computational tasks across a cluster of multiple machines, known as nodes. Each node contributes its processing power and memory, allowing Spark to handle datasets that are too large to fit on a single computer.

Rather than storing and processing all data on one machine, Spark divides the data into smaller partitions and distributes them across the nodes in the cluster. Each node works independently on its assigned portion of the data, performing a subset of the overall computations. This approach allows data processing and calculations to occur in parallel, significantly reducing execution time for many data-intensive tasks.

**Why is Spark Fast?**

One of Spark's major advantages is its ability to perform computations in memory whenever possible, reducing the need for slower disk operations. Combined with parallel processing, this makes Spark much faster than traditional big data frameworks for many workloads.

Suppose you have a dataset containing one billion customer transactions. Processing this dataset on a single computer could take hours. With Spark, the dataset can be divided into smaller chunks and distributed across dozens or even hundreds of nodes. Each node processes its assigned transactions simultaneously, and Spark then combines the results. This parallel execution can dramatically improve performance and scalability.

**Spark is particularly well-suited for:**

+ Large-scale data processing and analytics

+ Machine learning applications

+ Real-time data streaming

+ Graph processing

+ Interactive data exploration

+ ETL (Extract, Transform, Load) pipelines

**When Should You Use Spark?**

While Spark provides substantial computational power, it also introduces additional complexity, such as cluster management, data partitioning, and distributed system considerations. Therefore, Spark is not always the best solution.

Before choosing Spark, consider the following questions:

+ Is my dataset too large to fit into the memory or storage of a single machine?

+ Can my computations be divided into independent tasks that can run simultaneously?

+ Do I need faster processing than a single computer can provide?
Will the benefits of distributed computing outweigh the added complexity and infrastructure costs?

+ If the answer to these questions is yes, Spark may be an excellent choice.

As a real-world example, consider a financial institution that needs to detect fraudulent credit card transactions from billions of records generated daily. Analyzing all transactions on a single server would be slow and potentially impractical. Using Spark, the transaction data can be distributed across a cluster, where multiple nodes analyze different subsets of transactions in parallel. This enables the institution to identify suspicious activities much faster and respond to potential fraud in near real time.

**1.1) Spark DataFrames and RDDs**

At the heart of Apache Spark lies the Resilient Distributed Dataset (RDD), Spark's fundamental distributed data structure. An RDD is a fault-tolerant collection of data that is partitioned across multiple nodes in a cluster, enabling parallel processing of large datasets. RDDs provide fine-grained control over distributed computations and form the foundation upon which many Spark features are built.

While RDDs are powerful and flexible, they operate at a relatively low level and can be cumbersome to use for everyday data analysis tasks. As a result, most Spark users work with DataFrames, a higher-level abstraction built on top of RDDs that offers a more intuitive and efficient way to manipulate data.

**What is a Spark DataFrame?**

A Spark DataFrame is conceptually similar to a table in a relational database or a spreadsheet. Data is organized into:

Columns, which represent variables or attributes
Rows, which represent individual observations or records

This tabular structure makes DataFrames easier to understand, explore, and analyze compared to RDDs.

**Why Use DataFrames Instead of RDDs?**

One of the major advantages of DataFrames is their built-in optimization capabilities. When performing data transformations and analyses, there are often many different ways to achieve the same result. Some approaches can be significantly more efficient than others.

With RDDs, developers are largely responsible for optimizing their code and execution strategy. In contrast, Spark DataFrames leverage Spark's Catalyst Optimizer, which automatically analyzes queries and determines the most efficient execution plan. This allows users to focus on solving business and analytical problems rather than worrying about low-level performance tuning.

**Additional benefits of DataFrames:**

+ Faster execution through query optimization
+ Simplified syntax and easier code maintenance
+ Integration with Spark SQL
+ Better support for structured and semi-structured data
+ Improved memory and resource management

**1.2) How to Use Spark in Python?**

Before you can start processing data with Spark, you need to establish a connection to a Spark cluster. A Spark cluster is a group of interconnected computers that work together to perform large-scale data processing tasks.

In a typical Spark environment, the cluster runs on multiple machines. One machine serves as the master node (also known as the driver), which is responsible for coordinating the work. The master manages data distribution, schedules tasks, and oversees the execution of computations. The remaining machines, called worker nodes (historically referred to as slaves), perform the actual processing tasks assigned by the master and return their results when completed.

When learning Spark, setting up and managing a distributed cluster can be unnecessarily complicated. Therefore, many educational environments simulate a cluster on a single machine. **1.6) Converting a Pandas DataFrame to a Spark DataFrame**
Spark makes it easy to move data from Pandas into a distributed computing environment. The createDataFrame() method of the SparkSession class converts a Pandas DataFrame into a Spark DataFrame.

> spark_df = spark.createDataFrame(pandas_df)

The resulting Spark DataFrame can immediately be used with Spark's DataFrame API for data manipulation and analysis. However, the DataFrame is not automatically registered in the Spark catalog, meaning it cannot be referenced directly in SQL queries.

To make the DataFrame accessible through Spark SQL, you must register it as a temporary view using createTempView():

> spark_df.createTempView("my_table")

This creates a temporary table that can be queried within the current SparkSession.

To avoid errors when a table with the same name already exists, Spark provides the createOrReplaceTempView() method:

> spark_df.createOrReplaceTempView("my_table")

This method creates a new temporary view if none exists, or replaces the existing one if it does. It is generally the preferred approach because it prevents conflicts caused by duplicate table names. Note that temporary views exist only within the SparkSession in which they were created and are automatically removed when the session ends.

```python
# Do some practice on DataFrame conversions
import pandas as pd
import numpy as np

# Create pd_temp
pd_temp = pd.DataFrame(np.random.random(10))

# Create spark_temp from pd_temp
spark_temp = my_spark.createDataFrame(pd_temp)

# Examine the tables in the catalog
print(my_spark.catalog.listTables())

# Add spark_temp to the catalog
spark_temp.createOrReplaceTempView("temp")

# Examine the tables in the catalog again
print(my_spark.catalog.listTables())
```

# **2) Data Transformations Using Spark DataFrame**

In this section, you'll explore how to use Spark DataFrame methods to carry out common data transformations.

**2.1) Column Operations**

One frequently used operation is creating or modifying columns with the .withColumn() method. This method accepts two parameters: the name of the column to create or update (as a string) and a Column object that defines the column's values.

Creating a Column object is straightforward. You can reference an existing column using syntax such as df.colName. Unlike pandas, Spark DataFrames are immutable, meaning they cannot be modified directly. As a result, methods like .withColumn() always return a new DataFrame rather than changing the existing one.

To retain the changes, you need to assign the result back to a variable, often the original DataFrame name:

> df = df.withColumn("newCol", df.oldCol + 1)

This statement generates a new DataFrame containing all the original columns along with a new column, newCol, whose values are calculated by adding 1 to the corresponding values in oldCol. If you want to replace an existing column instead of creating a new one, simply provide the existing column’s name as the first argument to .withColumn().

```python
flights = my_spark.read.csv('flights_data.csv', header=True)
flights.show()

flights.name = flights.createOrReplaceTempView('flights')
my_spark.catalog.listTables()

# Create the DataFrame flights
flights_df = my_spark.table('flights')

# Show the head
print(flights_df.show())

# include a new column called duration_hrs
flights = flights.withColumn('duration_hrs', flights.air_time / 60)
flights.show()
```

**2.2) Filtering Data**

The .filter() method in Spark is used to select rows that meet a specific condition, similar to the WHERE clause in SQL. It accepts either a Spark Column expression that evaluates to True or False, or a SQL-style condition provided as a string.

For example, the following two statements are equivalent and return the same results:

> flights.filter(flights.air_time > 120).show()

> flights.filter("air_time > 120").show()

In the first statement, the filtering condition is expressed using a Spark Column object. In the second, the same condition is written as a SQL-like string. Both approaches keep only the rows where air_time is greater than 120.

```python
# Filter flights with a SQL string
long_flights1 = flights.filter('distance > 1000')

# Filter flights with a boolean column
long_flights2 = flights.filter(flights.distance > 1000 )

# Print the data to check they're equal
long_flights1.show()
long_flights2.show()
```

**2.3) Column Selection**

The Spark equivalent of SQL’s SELECT statement is the .select() method. This method accepts one or more arguments, each representing a column you want to keep in the resulting DataFrame. You can specify columns either by their names as strings or as Column objects using the df.colName syntax.

When working with Column objects, you can also apply transformations such as arithmetic operations, similar to what you can do with .withColumn(). This allows you to modify the selected data while retrieving it.

A key distinction between .select() and .withColumn() is the structure of the DataFrame they return. The .select() method produces a DataFrame containing only the columns you specify, whereas .withColumn() returns all existing columns along with the newly created or modified column.

Because carrying unnecessary columns can increase the amount of data being processed, it is often beneficial to remove unneeded columns early in your workflow. In situations where you only need a subset of columns, .select() is typically the more appropriate choice.

```python
# Select the first set of columns as a string
selected_1 = flights.select('tailnum', 'origin', 'dest')

# Select the second set of columns using df.col_name
temp = flights.select(flights.origin, flights.dest, flights.carrier)

# Define first filter to only keep flights from SEA to PDX.
FilterA = flights.origin == 'SEA'
FilterB = flights.dest == 'PDX'

# Filter the data, first by filterA then by filterB
selected_2 = temp.filter(FilterA).filter(FilterB)

selected_2.show()
```

Just like in SQL, Spark’s .select() method can be used not only to retrieve columns but also to perform transformations on them. When you reference a column using the df.colName syntax, you can apply operations directly within the .select() statement, and the resulting DataFrame will contain the transformed values.

For example:

> flights.select(flights.air_time / 60)

This expression returns flight durations converted from minutes to hours.

You can also rename transformed columns using the .alias() method. For instance, if you want the resulting column to be named duration_hrs, you can write:

> flights.select((flights.air_time / 60).alias("duration_hrs"))

Spark also provides the .selectExpr() method, which allows you to express the same transformations using SQL syntax. The following statement produces the same result:

> flights.selectExpr("air_time/60 as duration_hrs")

In this context, SQL’s AS keyword serves the same purpose as the .alias() method. When selecting multiple columns with .selectExpr(), simply provide multiple SQL expressions as separate string arguments.

```python
# Create a table of the average speed of each flight both ways.
# Calculate average speed by dividing the distance by the air_time (converted to hours).
# Use the .alias() method name

# Define avg_speed
avg_speed = (flights.distance/(flights.air_time/60)).alias("avg_speed")

# Select the correct columns
speed_1 = flights.select('origin','dest','tailnum', avg_speed)

# Create the same table using a SQL expression
speed_2 = flights.selectExpr('origin','dest','tailnum','distance/(air_time/60) as avg_speed')
speed_2.show()
```

# **2.4) Data Aggregation**
Many commonly used aggregation functions in Spark, such as .min(), .max(), and .count(), are methods of the GroupedData class rather than the DataFrame itself. A GroupedData object is created by calling the .groupBy() method on a DataFrame.

To use an aggregation function, first apply .groupBy() to your DataFrame and then call the desired aggregation method. For example, to determine the minimum value in a column named col within a DataFrame called df, you can write:

> df.groupBy().min("col").show()

In this example, .groupBy() creates a GroupedData object, enabling the use of the .min() method. The minimum value of col is then calculated and returned as a new DataFrame, which is displayed using .show().

```python
from pyspark.sql.functions import when, col
from pyspark.sql.types import FloatType

# Ensure flights DataFrame is in a state where these columns are strings before conversion
# If this cell is run independently, `flights` might not be the initial DataFrame.
# Reloading from CSV or ensuring it's the correct starting state is good practice.
# Assuming `flights` is loaded with string columns if `inferSchema=False` was used, or inferSchema=True sometimes incorrectly infers string if 'NA' is present.
# For robustness, we will ensure that the current `flights` DataFrame is used, and conversions are applied.

# Handle 'NA' values and cast 'distance' to FloatType
flights = flights.withColumn('distance', when(col('distance') == 'NA', None).otherwise(col('distance')).cast(FloatType()))

# Handle 'NA' values and cast 'air_time' to FloatType
flights = flights.withColumn('air_time', when(col('air_time') == 'NA', None).otherwise(col('air_time')).cast(FloatType()))

flights.describe('air_time', 'distance').show()

# Find the length of the shortest (in terms of distance) flight that left PDX
flights.filter(flights.origin =='PDX').groupBy().min('distance').show()

# Find the length of the longest (in terms of time) flight that left SEA
flights.filter(flights.origin == 'SEA').groupBy().max('air_time').show()

# get the average air time of Delta Airlines flights  that left SEA.
flights.filter(flights.carrier == 'DL').filter(flights.origin == 'SEA').groupBy().avg('air_time').show()

# get the total number of hours all planes in this dataset spent in the air by creating a column called duration_hrs
flights.withColumn('duration_hrs', flights.air_time/60).groupBy().sum('duration_hrs').show()
```

One of the key strengths of aggregation is the ability to perform calculations within groups of data. In PySpark, grouped operations are handled through the pyspark.sql.GroupedData class, which is created when you apply the .groupBy() method to a DataFrame.

Previously, you used .groupBy() without any arguments to perform aggregations across an entire DataFrame. However, when you provide one or more column names to .groupBy(), Spark organizes the data into groups based on the values in those columns. Aggregation functions then operate on each group separately, producing results similar to those generated by a SQL GROUP BY clause.


```python
# Group by tailnum colum
by_plane = flights.groupBy('tailnum')

# Use the .count() method with no arguments to count the number of flights each plane made
by_plane.count().show()

# group by origin column
by_origin = flights.groupBy('origin')

# Find the .avg() of the air_time column to find average duration of flights from PDX and SEA
by_origin.avg('air_time').show()
```

Alongside the GroupedData methods you've already encountered, Spark also provides the .agg() method for performing more flexible aggregations. This method allows you to supply one or more aggregate column expressions, which can make use of any of the functions available in the pyspark.sql.functions submodule.

The pyspark.sql.functions submodule includes a wide range of built-in functions, such as those for calculating standard deviation and other statistical measures. Each of these aggregation functions is designed to operate on column names within a GroupedData object, enabling powerful and customizable summary computations.

```python
# Import pyspark.sql.functions as F
import pyspark.sql.functions as F

# Group by month and dest
by_month_dest = flights.groupBy('month', 'dest')

#convert to dep_delay to numeric column
flights = flights.withColumn('dep_delay', flights.dep_delay.cast('float'))

# Average departure delay by month and destination
#by_month_dest.avg('dep_delay').show()
```

**2.5) Joining DataFrames**

In PySpark, you can combine DataFrames using the .join() method. This method requires three main arguments. The first is the second DataFrame you want to merge with the original one. The second, on, specifies the key column(s) used for matching rows, provided as a string. These key columns must have the same name in both DataFrames. The third argument, how, determines the type of join to perform. In this course, the join type will consistently be set to how="leftouter".

The workspace already includes the flights dataset as well as a second dataset called airports.

```python
airports.show()

# Rename the faa column
airports = airports.withColumnRenamed('faa','dest')

# Join the DataFrames
flights_with_airports = flights.join(airports, on='dest', how='leftouter')
flights_with_airports.show()
```
